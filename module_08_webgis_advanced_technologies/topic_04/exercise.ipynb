## Objective and completed work

This submission extends the official Module 08 Leaflet/Folium examples into a deployable WebGIS that integrates **raster layers, vector layers, and sensor-network data**. The client uses Leaflet and responsive CSS; Flask exposes GeoJSON/CSV data through API endpoints; Docker provides a reproducible open-source deployment.

The implementation is stored in [`webgis_server/`](webgis_server/).


## Architecture

- **Raster:** OpenStreetMap and Esri World Imagery web tile layers.
- **Vector:** local flood-hazard polygons in GeoJSON.
- **Sensors:** local sensor-station GeoJSON plus CSV observation history.
- **Server:** Flask routes `/api/health`, `/api/sensors`, `/api/hazards`, `/api/sensor-history`.
- **Responsive design:** CSS media queries switch the dashboard between desktop and mobile layouts.
- **Open-source deployment:** MIT license, `requirements.txt`, and `Dockerfile`.
- **Static fallback:** if Flask API routes are unavailable, the browser loads the local files directly, so the HTML can also be demonstrated with static hosting.

> The included station/hazard data are demonstration data for coursework, not operational warnings.


````python
from pathlib import Path
import json

project = Path('webgis_server')
with (project / 'data' / 'sensors.geojson').open(encoding='utf-8') as f:
    sensors = json.load(f)
with (project / 'data' / 'hazard_zones.geojson').open(encoding='utf-8') as f:
    hazards = json.load(f)

print('Sensor stations:', len(sensors['features']))
print('Hazard polygons:', len(hazards['features']))
assert len(sensors['features']) >= 3
assert len(hazards['features']) >= 1
````

## Run the WebGIS server

