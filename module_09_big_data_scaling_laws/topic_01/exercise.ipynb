Built-up areas should generally form the warmest zones because impervious surfaces store heat and provide little evapotranspirative cooling. Irrigated agricultural areas are expected to be cooler, although dry or harvested fields may warm substantially. Water bodies and forested highlands should form the coolest zones because of thermal inertia, canopy shading, and evapotranspiration. NDVI provides a useful cross-check: high-NDVI areas should usually show lower LST, while low-NDVI built or bare surfaces should be warmer. This inverse LST–NDVI relationship can support urban heat-island identification, green-infrastructure planning, and heat-health preparedness. LST is surface skin temperature and should not be interpreted as air temperature.
