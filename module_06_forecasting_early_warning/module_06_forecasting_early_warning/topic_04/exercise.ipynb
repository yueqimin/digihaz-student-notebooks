## Assessment summary

### Components
- real-time rainfall and river-water-level monitoring;
- forecast/model and ICT infrastructure under PRABN;
- Normal / Alert / Warning / Danger decision levels;
- web warning dissemination, selected email/SMS subscriptions, and siren telemetry;
- community protective action and evacuation response.

### Main strengths
- nationwide real-time monitoring includes Penang and Kedah;
- clear severity terminology linked to preparedness/evacuation;
- multi-layer monitoring-model-dissemination architecture;
- multiple potential notification channels.

### Main weaknesses
- forecast-alert subscription coverage is not equivalent to nationwide monitoring coverage;
- last-mile receipt and comprehension are not guaranteed;
- static water-level thresholds do not directly encode local impacts;
- sensor/network failures and timestamp/data-quality problems can affect reliability;
- community outcome evidence needs post-event surveys and drills.


## Recommended improvements

1. Expand validated forecast-alert POI coverage and geofenced subscriptions.
2. Use redundant, multilingual last-mile channels (web/app + SMS/cell broadcast + sirens/radio/local relays).
3. Move from stage-only thresholds toward impact-based warnings tied to inundation, roads and critical facilities.
4. Add automated quality-control gating, failover gauges, backup communications and power.
5. Back-test the complete operational chain on multiple historical floods and track POD, FAR, CSI, lead time and P95 latency.
6. Measure warning receipt, understanding, trust and time-to-action at community level.
